```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

<h2>Swiggy Data 1</h2>


```python
#==========================
# Swiggy Data 1
#==========================

data = "/content/swiggy.xlsx"

df = pd.read_excel(data)
df.head()
```

```python
print("Number of Restaurant by Cities")
total = df["City"].value_counts()
print(total)
print()
plt.figure(figsize=(10,6))
plt.bar(total.index,total.values)
plt.title("Number Of Restaurant By Cities :")
plt.xlabel("City")
plt.ylabel("Count Of Restaurant ")
plt.show()
```

```python
print("Count of Rating Rate")
rating = df["Avg ratings"].value_counts()
print(rating)
print()
```

```python
print("Count of Rating Rate")
rating = df["Avg ratings"].value_counts()

print()
plt.figure(figsize=(10,6))
plt.scatter(rating.index,rating.values)
plt.title("Frequency of Rating Rate ")
plt.xlabel("Rate")
plt.ylabel("Count Of Rating Rate ")
plt.show()
```

```python
print()
plt.figure(figsize=(15,6))
plt.scatter(df["Price"],df["Avg ratings"])
plt.title("Price vs Rate ")
plt.xlabel("Price")
plt.ylabel("Rating Rate ")
plt.show()
```

```python
print()
plt.figure(figsize=(15,6))
plt.bar(df['Delivery time'],df["Avg ratings"])
plt.title("Delivery time vs Rate ")
plt.xlabel("Delivery time")
plt.ylabel("Rating Rate ")
plt.show()
```

```python
print(" City with Fastest delivery ")
top10_city = df.nsmallest(10, "Delivery time")

print()
plt.figure(figsize=(10,6))
sns.barplot(data=top10_city, x = "City", y="Delivery time")
plt.title("City with Fastest delivery ")
plt.xlabel("City")
plt.ylabel("delivery Time")
plt.show()

```

<h2> Swiggy Data 2</h2>

```python
#==========================
# Swiggy Data 2
#==========================

data2 = "/content/swiggy_file .xlsx"

df = pd.read_excel(data2)
df.head()
```

```python
print("Highest Rating By Cuisine")

top10_Cuisine = df.nlargest(10, "Number of ratings")

plt.figure(figsize=(12, 8))

sns.barplot(
    data=top10_Cuisine,
    x="Number of ratings",
    y="Cuisine"
)

plt.title("Cuisine with Highest Number of Ratings")
plt.xlabel("Number of Ratings")
plt.ylabel("Cuisine")

plt.show()
```
